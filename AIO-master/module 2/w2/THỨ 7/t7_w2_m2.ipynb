# database normalizaton 
### Chuẩn hóa cơ sở dữ liệu là gì?
- Là quá trình tổ chức dữ liệu trong cơ sở dữ liệu để:
- Giảm dư thừa dữ liệu
- Ngăn ngừa các bất thường trong thao tác dữ liệu
- Cải thiện tính toàn vẹn dữ liệu

### terminology 
- primary key 
- super key 

### super key 
- Siêu khóa là một tập hợp một hoặc nhiều cột mà:
- Xác định duy nhất mỗi hàng trong bảng
- Có thể chứa thêm các cột không cần thiết cho tính duy nhất

### nomarlization form 
#### 1NF : 
- Mỗi ô chỉ chứa một giá trị đơn (atomic values)
- Không có nhóm dữ liệu lặp lại
- Mỗi bảng có khóa chính

#### 2NF : 
- Đã đạt 1NF
- Mọi thuộc tính không khóa phải phụ thuộc đầy đủ vào khóa chính
- Loại bỏ phụ thuộc bộ phận (partial dependency)

#### 3NF : 
- Đã đạt 2NF
- Không có thuộc tính không khóa phụ thuộc bắc cầu vào khóa chính
- Loại bỏ phụ thuộc bắc cầu (transitive dependency)

#### BCNF : 
- Đã đạt 3NF
- Với mọi phụ thuộc hàm X → Y, thì X phải là siêu khóa



### SQL joins 
- inner join : lấy phần thuộc cả 2 bảng 
- truy vấn trên nhiều bảng kết hợp với "on" 

- self join : tự lấy dữ liệu từ chính bảng đó 

- joining mutiple tables : lấy dữ liệu từ nhiều bảng 

- compound join condition : join dựa trên nhiều hơn 1 khoá chính 

- implicit join syntax : 

- cross join : kết hợp từng trường của bảng bên trái kết hợp với từng trường bên bảng bên phải  

- outer join : left join , right join 

### using clause 
- thay vì viết tường minh là "c.customer_id = o.customer_id" thì viết using(customer_id)
