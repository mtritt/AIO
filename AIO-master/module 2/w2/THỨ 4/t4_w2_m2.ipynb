# einstein summation : tương tự tích vô hướng 
- Einstein summation (quy ước tổng Einstein) là một cách viết cực kỳ gọn gàng và mạnh mẽ để biểu diễn các phép toán trên mảng/tensor nhiều chiều. Nó loại bỏ hoàn toàn các ký hiệu tổng (Σ) và dựa vào việc lặp lại chỉ số.
- Quy tắc vàng của Einstein summation: "Khi một chỉ số xuất hiện hai lần trong một biểu thức, nó ngầm định một phép tổng theo chỉ số đó."


- Cú pháp của np.einsum: np.einsum('subscripts', operand1, operand2, ...)

- Chỉ số đầu vào: Mô tả kích thước của các toán hạng

- Chỉ số đầu ra: Mô tả kích thước kết quả (sau dấu ->)

- Quy tắc:

- Chỉ số xuất hiện một lần trong đầu vào và có trong đầu ra → giữ nguyên chiều đó

- Chỉ số xuất hiện hai lần trong đầu vào và không có trong đầu ra → tổng theo chiều đó



```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Ví dụ 1: Tích vô hướng    
result = np.einsum('i,i->', a, b)
# ↑ Chỉ số 'i' xuất hiện 2 lần → tổng theo i 
# Tính: 1×4 + 2×5 + 3×6 = 4 + 10 + 18 = 32 
result 
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Ví dụ 2: Nhân ma trận
result = np.einsum('ij,jk->ik', A, B)
# ↑ 'j' xuất hiện 2 lần → tổng theo j (bị loại khỏi đầu ra)
# ↑ 'i' và 'k' chỉ xuất hiện 1 lần → được giữ lại trong đầu ra
result 
```
