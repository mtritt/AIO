```python
# matrix - matrix : tương tự nhân 2 ma trận 
A = np.array([[1 , 2] , [3 , 4]])
B = np.array([[1 , 2] , [3 , 4]])
A.dot(B) 
# A dot B != B dot A 
```

```python
# determinant for matrix square 
np.linalg.det(A)
```

```python
"""
    x.A = B ==> x.A.A^-1 = B.A^-1 ==> x = B.A^-1
    if det(A) != 0 : invertible (khả nghịch) ==> có khả năng tìm được x 
        B = A^-1 = 1/det(A) * [[d , -b] , [-c , a]] và đan dấu 
    else : 
        A is not invertible 
"""
```

```python
# matrix khả nghịch (nghịch đảo)
np.linalg.inv(A) 
```

# cosine similarity 
- measure the similarity between two vectors ==> góc giữa 2 vector 

```python
A = np.array([1 , 2 , 3])
B = np.array([4 , 5 , 6])
```

```python
cosine = np.dot(A , B)/(np.linalg.norm(A) * np.linalg.norm(B))
cosine 
```

- measure the similarity between two images 

```python
import numpy as np 
from PIL import Image 

image1 = Image.open('/Users/macos/AIO/module 2 /w2/THỨ 3/data/image1.jpg')
image2 = Image.open('/Users/macos/AIO/module 2 /w2/THỨ 3/data/image2.jpg')
query = Image.open('/Users/macos/AIO/module 2 /w2/THỨ 3/data/query_image.jpg')
```

```python
img1_data = np.array(image1)
img2_data = np.array(image2)
query_data = np.array(query)
```

```python
img1_gray = np.sum(img1_data , axis = -1 , )/3
img2_gray = np.sum(img2_data , axis = -1 , )/3
query_gray = np.sum(query_data , axis = -1 , )/3
```

```python
img1_flatten = img1_gray.flatten() # duỗi từ ma trận về vector 
img2_flatten = img2_gray.flatten() # duỗi từ ma trận về vector 
query_flatten = query_gray.flatten() # duỗi từ ma trận về vector 
```

```python
def compute_consine_similary(A , B) : 
    cosine = np.dot(A , B)/(np.linalg.norm(A) * np.linalg.norm(B))
    return cosine 
```

```python
cosine1 = compute_consine_similary(img1_flatten , img2_flatten)
cosine1
```

```python
cosine2 = compute_consine_similary(img2_flatten , query_flatten)
cosine2 
```
