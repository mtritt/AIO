## eigenvector and eigenvalue (vector riêng và giá trị riêng )
- vector riêng và giá trị riêng 
- Eigenvector là vector v ≠ 0 mà khi nhân với ma trận A, nó không thay đổi hướng, chỉ thay đổi độ dài.
- Eigenvalue (λ) là hệ số mà eigenvector được nhân lên.

- eigenvector : những hướng "đặc biệt" mà khi biến dạng , không gian chỉ co dãn theo hướng đó , không bị xoay hay bẻ cong 
- eigenvalue : mức độ co giãn theo hướng đó 

```python
import matplotlib.pyplot as plt

# Minh họa bằng hình ảnh
def visualize_eigen():
    A = np.array([[2, 1], 
                  [1, 2]])
    
    eigenvalues, eigenvectors = np.linalg.eig(A)
    
    # Vector gốc
    original_vectors = np.array([[1, 0], [0, 1]])
    
    # Vector sau biến đổi
    transformed_vectors = A @ original_vectors
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
    
    # Trước biến đổi
    ax1.quiver([0, 0], [0, 0], original_vectors[0], original_vectors[1], 
               angles='xy', scale_units='xy', scale=1, color=['r', 'b'])
    ax1.set_xlim(-3, 3)
    ax1.set_ylim(-3, 3)
    ax1.grid(True)
    ax1.set_title('Trước biến đổi')
    ax1.set_aspect('equal')
    
    # Sau biến đổi
    ax2.quiver([0, 0], [0, 0], transformed_vectors[0], transformed_vectors[1], 
               angles='xy', scale_units='xy', scale=1, color=['r', 'b'])
    # Vẽ eigenvectors
    for i in range(len(eigenvalues)):
        ev = eigenvectors[:, i] * eigenvalues[i]
        ax2.quiver(0, 0, ev[0], ev[1], angles='xy', scale_units='xy', scale=1, 
                  color='g', alpha=0.7, linestyle='--')
    
    ax2.set_xlim(-3, 3)
    ax2.set_ylim(-3, 3)
    ax2.grid(True)
    ax2.set_title('Sau biến đổi (eigenvectors màu xanh lá)')
    ax2.set_aspect('equal')
    
    plt.tight_layout()
    plt.show()

visualize_eigen()
```

```python
# tìm eigenvalues và eigenvectors 
def compute_eigenvalues_eigenvector(matrix) : 
    eigenvalues , eigenvectors = np.linalg.eig(matrix)
    return eigenvalues , eigenvectors 

matrix = np.array(
    [[3 , 4 , 5], 
    [3 , 6 , 8] , 
    [9 , 5 , 3]]
)

eigenvalues , eigenvectors = compute_eigenvalues_eigenvector(matrix)
print(eigenvalues)
print(eigenvectors)
```

```python
# cosin similarity 
def compute_cosin(v1 , v2): 
    # tích vô hướng 
    dot_product = np.dot(v1 , v2)

    # tích độ dài 
    norm_v1 = np.linalg.norm(v1)
    norm_v2 = np.linalg.norm(v2)

    # tránh việc chia cho 0 
    if norm_v1 == 0 or norm_v2 == 0 : return 0 


    return dot_product / (norm_v1 * norm_v2)

vector_1 = np.arange(9)
vector_2 = np.arange(9)

compute_cosin(vector_1 , vector_2)
```

## background subtraction (tách nền)
- so sánh ảnh gốc với nền thật 
- tách đối tượng tiền cảnh 
- thay thế nền thật bằng nền giả 
- _, difference_binary = cv2.threshold(difference, 15, 255, cv2.THRESH_BINARY) : ảnh đầu vào ( sự khác biệt giữa 2 ảnh , ngưỡng , giá trị max , thresbinary : cách xử lý)

- dấu '_,' là quy ước trong python chỉ các biến không sử dụng 
- hàm threshold trả về 2 giá trị : retval là giá trị ngưỡng được sử dụng , dst là ảnh kết quả sau khi áp dụng threshold 


- hàm threshold hoạt động : 
- if pixel_value > 15:      → gán thành 255 (TRẮNG - đối tượng)
- else:                     → gán thành 0   (ĐEN - nền)

- bản chất threshold trả về 2 giá trị đó là : retval , dst 
- nhưng ta chỉ cần 1 giá trị đó là dst nên sử dụng '_,' để output chỉ có dst 

```python
import numpy as np 
import cv2 

# đọc ảnh nền thật 
img_bg = '/Users/macos/AIO/module 2 /w2/THỨ 8/Solution/Exercise04_Data/GreenBackground.png'
bg = cv2.imread(img_bg, 1)
bg = cv2.resize(bg, (640, 480))

# đọc ảnh nền có đối tượng 
obj_img = '/Users/macos/AIO/module 2 /w2/THỨ 8/Solution/Exercise04_Data/Object.png'
obj = cv2.imread(obj_img, 1)
obj = cv2.resize(obj, (640, 480))  # SỬA: dùng obj thay vì obj_img

# tính sự khác biệt và tránh số âm 
difference = cv2.absdiff(bg, obj)

# tạo ra threshold 
_, difference_binary = cv2.threshold(difference, 15, 255, cv2.THRESH_BINARY)

# đọc ảnh nền giả 
new_bg = '/Users/macos/AIO/module 2 /w2/THỨ 8/Solution/Exercise04_Data/NewBackground.jpg'
new_img = cv2.imread(new_bg, 1)
new_img = cv2.resize(new_img, (640, 480))  # SỬA: dùng new_img thay vì new_bg

# thay thế nền
res = np.where(difference_binary == 0, new_img, obj)  # SỬA: dùng new_img thay vì new_bg
cv2.imwrite('output.png', res)
```
