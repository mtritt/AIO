# Image depth estimation 

## introduction 

- Stereo Matching: A computer vision technique used to determine the depth of objects in a scene by comparing two or more images taken from slightly different viewpoints.
By analyzing the disparity between corresponding points in these images, stereo matching algorithms estimate the distance to various points, creating a 3D representation of the scene.



- stereo matching applications : 3D ,...

- disparity map : sự chênh lệch pixel giữa 2 ảnh

```python
import numpy as np 
import cv2 

def l1_distance(x , y) : 
    return abs(x - y)

def l2_distance(x , y):
    return (x - y) ** 2 
```

```python
# read image then convert to graysalce 
left = cv2.imread(left_img , 0)
right = cv2.imread(right_img , 0)

# mặc định python sẽ đọc ảnh kiểu uint8 nên phải đổi sang float32 
left = left.astype(np.float32)
right = right.astype(np.float32)

# (h , w)
height , width = left.shape[:2]

# create blank deparity map 
depth = np.zeros((height , width) , np.uint8)
scale = 10 
max_value = 255 
disparity_range = 16 
```

```python
for y in range(height) : 
    for x in range(width) : 
        # find y where cost have minimum value 
        disparity = 0 
        cost_min = max_value 
        # dịch từ 0 đến pixel 15 để so sánh 
        for j in range(disparity_range) : 
            # (x - j) < 0 nếu vượt ngoài biên ảnh
            # 
            cost = max_value if (x - j) < 0 else l1_distance(int(left[y , x]) , int(right[y , x - j]))
            if cost < cost_min : 
                cost_min = cost  
                disparity = j 
        depth[y , x] = disparity * scale 
```
