# SQL 
## introduce SQL  
- CRUD : create - read - update - delete 

- create database : create database [name]
- delete database : drop database [name]


- create table : create table [name]
- auto_increment : tự động tăng 

- insert data : insert into [table_name] (col1 , col2 , col3 ,....) values ()

- query data : select col1 , col2 ,... from table 

## how to use SQL for data analys 
- SQL (structer query language) is a powerful tool for data analys . It allows you to query and manipulate databases efficiently 

#### select data from table 
- select * from [name_table]
- select [name_col] from [name_table]

#### alias 
- select [name_col] as [alias] from [table_name]
- ==> tạo ra 1 cột có tên như đã đặt sau alias 
- có thể sử dụng "select distict " để loại bỏ giá trị trùng lặp 

#### where 
- select [col_name] from [table_name] where [condition]

#### in - between 
- in giống như so sánh nhưng ngắn hơn 

#### is NULL - order by - limit 
- is NULL : check rỗng  
- order by : sắp xếp tăng dần 
- order by [col_name] desc : sắp xếp giảm dần

- like : như so sánh "=="
