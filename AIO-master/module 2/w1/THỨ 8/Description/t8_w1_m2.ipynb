```python
# ex2 
import cv2 
import numpy as np 
import matplotlib.pyplot as plt 

dog_url = '/Users/macos/AIO/module 2 /w1/THỨ 8/Description/dog.jpeg'
img = cv2.imread(dog_url , 1)
img = img[: , : , ::-1]
plt.imshow(img)

""" cách 1 : tính trung bình """
def compute_avg(img) :
    return np.sum(img) / 3 

img = np.apply_along_axis(compute_avg , axis = 2 , arr = img)
img = img.astype(np.uint8)
plt.imshow(img , cmap = 'gray')

"""cách 2 : Luminosity method """
def compute_luminosity(img) :
    return (0.21 * img[0] + 0.72 * img[1] + 0.07 * img[2])

img = np.apply_along_axis(compute_luminosity , axis = 2 , arr = img)
img = img.astype(np.uint8)
plt.imshow(img , cmap = 'gray')

"""cách 3 : Weighted method """
def compute_weighted(img) : 
    return (np.max(img , axis = 2) + np.min(img , axis = 2))/2

img = np.apply_along_axis(compute_weighted , axis = 2 , arr = img)
img = img.astype(np.uint8)
plt.imshow(img , cmap = 'gray')
```

```python
# ex3 
import pandas as pd 
import numpy as np

df = pd.read_csv('advertising.csv')
data = df.to_numpy()

# get the maximum value of the sales column and its corresponding index 
sales_data = data[: , 3]
sales_max = np.max(sales_data)
sales_index = np.argmax(sales_data) # lấy index 
data[175]
```

```python
tv_data = data[: , 0]
tv_mean = np.mean(tv_data)
tv_mean 
```

```python
# sum vừa để tính tổng cũng vừa để đếm 
sales_res = np.sum(data[: , 3] >= 20)
sales_res 
```

```python
sales_data = data[: , 3]
sales_cond = sales_data >= 15

radio_data = data[: , 1]
radio_cond = radio_data * sales_cond
sales_mean = np.sum(radio_cond) / np.sum(sales_cond)
sales_mean 
```

```python
sales_data = data[: , 3]
sales_mean = sales_data.mean()
np.where(sales_data < sales_mean , "Bad" ,np.where(sales_data > sales_mean , "Good" , "Average"))
```
