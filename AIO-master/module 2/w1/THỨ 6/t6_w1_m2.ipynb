```python
import numpy as np
import glob # hỗ trợ đọc file trong folder 
from tqdm import tqdm # hiển thị thời gian chạy 

paths = glob.glob('samples/*.jpg').sort()
paths = sorted(paths)
# sorted tạo ra list mới và sắp xếp list đó 
# sort sắp xếp list gốc 

# read , resize and add to list 
res = []
for path in tqdm(paths) : 
    img = cv2.imread(path , 1)
    img = cv2.resize(img , (100 , 100))
    res.append(img)

res = np.array(res)
print(res.shape)
```

## np.clip() function 
- giới hạn các giá trị trong mảng numpy trong 1 khoảng nhất định 
- syntax : np.clip(a, a_min, a_max, out=None)
- np.clip() đảm bảo tất cả pixel đều nằm trong khoảng hợp lệ [0, 255] trước khi chuyển đổi sang uint8.


```python
# numpy.clip() 
# các số trong image được biểu diễn từ 0 - 255 và có tính chất xoay vòng 
import numpy as np 
import cv2 
import matplotlib.pyplot as plt 

img = cv2.imread('image.png' , 1)
img = img.astype(float)
print(img)

# process
img = img + 20 
img = np.clip(img , a_min = 0 , a_max = 255)

img = img[: , : ,::-1]
img = img.astype(np.uint8)
plt.imshow(img)
```

```python
# where function 
arr = np.arange(10)
print(arr)

out = np.where(arr < 3 , arr , arr * 2)
# nếu đúng nhận tham số 1 , nếu sai nhận tham số 2 
print(out)
```

```python
# numpy.clip() 
# các số trong image được biểu diễn từ 0 - 255 và có tính chất xoay vòng 
import numpy as np 
import cv2 
import matplotlib.pyplot as plt 

img = cv2.imread('image.png' , 1)
img = img.astype(float)

# process
img = img + 20 
img = np.where(img , 0 , img)

img = img[: , : ,::-1]
img = img.astype(np.uint8)
plt.show(img)
```

## color conversion 
- apply_along_axis() function 

```python
def compute_avg(vector):
    return vector.mean()

data = np.arange(9).reshape(3 , 3)

vector = np.apply_along_axis(compute_avg , axis = 0 , arr = data)

print(vector)
```

```python
import numpy as np 
import cv2 
import matplotlib.pyplot as plt 

def compute_avg(vector) :
    return vector.mean()

img = cv2.imread('image.png' , 1)
img = np.apply_along_axis(compute_avg , axis = 2 , arr = img)

# show 
img = img.astype(np.uint8)
plt.imshow(img , cmap = 'gray')
```
