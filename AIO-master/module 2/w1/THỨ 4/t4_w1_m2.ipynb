# introduction to numpy 


```python
import numpy as np 

data = np.array([1 , 2 , 3])
print(data.shape) # (3.) ==> mảng 1 chiều có 3 phần tử 
```

```python
def multiply(number) : 
    return number * 2 

multiply_arr = np.vectorize(multiply)
data = [1 , 2 , 3]
data_np = np.array(data)
print(multiply_arr(data_np))

```

```python
data = [2 , 4 , 6 , 8]
data_np = np.array(data)
data_np = data_np * 2 
print(data_np)
```

```python
shape = (10 , 10)
data = np.zeros(shape)
print(data)
arr_one = np.ones((4 , 5)) 
arr_one = arr_one + 1
print(arr_one)
```

```python
data = np.eye(3) # ma trận đơn vị 
print(data)
```

```python
data = np.array([1 , 4])
data = np.repeat(data , 2) # nhân phần tử lên 2 lần 
# axis dành cho dl nhiều chiều 
# axis-0 đi xuống 
# axis-1 đi lên 
print(data)
```

```python
data = np.arange(100).reshape(10 , 10)
print(data.shape)

data = np.repeat(data , 2 , axis=0).repeat(2 , axis=1)
print(data)
```

# array indexing 

```python
data = np.arange(9).reshape(3 , 3)
print(data)

sub_data = data[1 , :] # [i , j]
print(sub_data)

# array dùng chung vùng nhớ nên nếu thay đổi ở đâu thì array gốc sẽ thay đổi 

```

```python
data = np.arange(9).reshape(3 , 3)
print(data)
# truy xuất nhiều hơn 1 phần tử 
# data[[list_i] , [list_j]]
print(data[[1 , 2] , [0 , 2]])
print(data > 4)
```

```python
data = np.arange(9).reshape(3 , 3)
print(data)
print(np.sum(data , axis = 0))
```

# broadcasting 
- ví dụ : nếu có ma trận 2 chiều nhân với 1 số thì số đó sẽ được nhân bản bằng với size của ma trận 2 ==> kết quả 

```python
data = np.array([1 , 2 , 3])
# print(data)

data_exp = np.exp(data)
data_sum = np.sum(data)
result = data_exp / data_sum
print(result)
print(data)
```
