```python
import numpy as np

# Dữ liệu mẫu
x = [7, 18, 29, 2, 10, 9, 9]
y = [1, 6, 12, 8, 6, 21, 10]

# Tính correlation matrix
corr_matrix = np.corrcoef(x, y)
print("Correlation matrix (2x2):")
print(corr_matrix)

# Lấy riêng hệ số tương quan giữa x và y
corr_xy = corr_matrix[0, 1]
print(f"\nCorrelation coefficient between x and y: {corr_xy:.4f}")

# Hoặc trực tiếp
corr_direct = np.corrcoef(x, y)[0, 1]
print(f"Direct access: {corr_direct:.4f}")
```

- hàm tính correlation trong numpy sẽ có output dạng 
- [[ corr(x, x), corr(x, y) ],
-  [ corr(y, x), corr(y, y) ]]

### template matching (khớp mẫu)

1. variance (phương sai)
- dùng để đo lường sự phân tán của 1 biến ngẫu nhiên xung quanh giá trị trung bình của nó 
- cho biết dữ liệu "tập trung" hay "phân tán" như thế nào 

2. covariance (hiệp phương sai)
- đo lường hướng của mối quan hệ tuyến tính giữa 2 biến 
- cho biết 2 biến cùng chiều hay ngược chiều 
- nhưng không cho biết độ mạnh , yếu của mối quan hệ , chỉ biết hướng 

3. correlation coefficent (hệ số tương quan)
- Đo lường cường độ và hướng của mối quan hệ tuyến tính giữa hai biến
- Chuẩn hóa covariance về khoảng [-1, 1]

4. correlation (tương quan)
- Khái niệm tổng quát về mối quan hệ giữa hai hay nhiều biến
- Có nhiều loại: Pearson (tuyến tính), Spearman (hạng), Kendall (tau)
- có thể phát hiện các mối quan hệ phi tuyến tính 


### các loại correlation coefficent 
1. Pearson Correlation Coefficient
- Đo lường mối quan hệ tuyến tính giữa hai biến liên tục
- điều kiện sử dụng : 
- Dữ liệu liên tục (continuous)
- Phân phối chuẩn (hoặc gần chuẩn)
- Quan hệ tuyến tính (linear relationship)
- Không có outliers mạnh
- Phương sai đồng nhất (homoscedasticity)
- hạn chế : 
- Không phát hiện quan hệ phi tuyến
- Rất nhạy với outliers

2. Spearman Rank Correlation
- Đo lường mối quan hệ đơn điệu (monotonic) giữa hai biến
- Không cần quan hệ tuyến tính, chỉ cần cùng hướng thay đổi
- Tính correlation trên hạng (rank) của dữ liệu, không phải giá trị gốc



- tuyến tính : có thể biểu diễn bằng 1 đường thẳng 
- phi tuyến : không thể biểu diễn bằng 1 đường thẳng , có thể là đường cong , parabol ,....
