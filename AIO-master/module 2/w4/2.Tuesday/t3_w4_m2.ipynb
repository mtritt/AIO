### discrete random variables
- random varibales : một biến ngẫu nhiên , thường được kí hiệu là X , Y , Z , .... là một hàm đo lường được từ không gian mẫu của một phép thử ngẫu nhiên vào tập số thực R 
---- 
- a continuous random variables (biến ngẫu nhiên liên tục) : Biến có tập giá trị có thể đếm được (hữu hạn hoặc vô hạn đếm được).
----
- discrete random variables (biến ngẫu nhiên rời rạc) : Biến có tập giá trị lấp đầy một hoặc nhiều khoảng trên trục số thực (không đếm được).

- probability mass function : p(x) = P(X = x) 
----
- probability distribution function (PDF) (hàm phân phối xác suất) : F(x) = P(X <= x)

- kỳ vọng (expectation) : E(X) 
- phương sai (variance) : var(X) 

- hiệp phương sai (covariance) : cov(X , Y) 
- hệ số tương quan (correlation) : corr(X , Y) 

### Cov 
- công thức : Cov(X,Y)=E[(X−E[X])(Y−E[Y])]
- Cov(X , Y) > 0 : quan hệ cùng chiều . Khi X tăng thì Y cũng tăng và ngược lại 
- Cov(X , Y) < 0 : quan hệ ngược chiều . Khi X tăng thì Y giảm và ngược lại 
- Cov(X , Y) = 0 : không có quan hệ tuyến tính 
- --> covariance chỉ cho biết hướng của mối quan hệ chứ không biết được cường độ của nó 

### Corr 
- Hệ số tương quan chuẩn hóa hiệp phương sai, đưa nó về một thang đo không đơn vị và giới hạn trong khoảng [-1, 1].
- độ lệch chuẩn là căn bậc hai của phương sai 

### quan hệ tuyến tính và phi tuyến tính 
- quan hệ tuyến tính : hai biến X và Y có quan hệ tuyến tính nếu khi vẽ chúng lên biểu đồ phân tán thì các điểm có xu hướng phân bố thành một đường thẳng , mối quan hệ có thể được mô tả (xấp xỉ) bằng một phương trình đường thẳng Y = aX + b + c 
- quan hệ phi tuyến tính : hai biến X và Y có quan hệ phi tuyến tính nếu mối quan hệ giữa chúng không thể được miêu tả đủ tốt bằng một đường thẳng . Các điểm được phân bố theo một đường cong hoặc hình dạng phức tạp
