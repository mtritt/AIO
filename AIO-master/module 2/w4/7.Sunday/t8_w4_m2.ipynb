```python
import numpy as np 
```

```python
def compute_mean(X):
    return np.mean(X)

X = [2 , 0 , 2 , 2,7 , 4 ,-2 , 5 , -1 , -1]
compute_mean(X)
```

```python
def compute_median(X): 
    size = len(X) 
    X = np.sort(X) 
    if size % 2 == 0 : 
        return (X[int(size / 2)] + X[int(size / 2 + 1)])/2
    else : 
        return X[int((size + 1)/2)]

X = [1 , 5 , 4 , 4 , 9 , 13]
compute_median(X)
```

```python
# độ lệch chuẩn - phương sai 
def compute_std(X): 
    mean = compute_mean(X) 
    variance = 0 
    for i in range(len(X)) : 
        variance += (X[i] - mean) ** 2
    variance = variance / len(X) 
    return np.sqrt(variance)

X = [ 171 , 176 , 155 , 167 , 169 , 182]
compute_std(X)
```

```python
# correlation coefficent 
def compute_correlation_coefficent(X , Y) : 
    N = len(X) 
    numerator = 0 
    denominator = 0 

    for i in range(N) : 
        numerator += X[i] * Y[i] 
    numerator = numerator * N - (np.sum(X) * np.sum(Y))

    square_X = np.sum(X ** 2)
    square_X = N * square_X - np.sum(X) ** 2

    square_Y = np.sum(Y ** 2)
    square_Y = N * square_Y - np.sum(Y) ** 2

    denominator = np.sqrt(square_X) * np.sqrt(square_Y) 

    return np.round(numerator / denominator , 2)

X = np . asarray ([ -2 , -5 , -11 , 6 , 4 , 15 , 9])
Y = np . asarray ([4 , 25 , 121 , 36 , 16 , 225 , 81]) 
compute_correlation_coefficent(X , Y) 
```
