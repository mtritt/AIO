## what is a PMF , PDF , CDF 
### probability distribution (phân phối xs)
----
1.PMF (probability mass function) - hàm khối lượng xác suất 
- hàm xs cho biến ngẫu nhiên rời rạc 
- xây dựng biểu đồ cột 
----
2.PDF (probability density function) - hàm mật độ xác suất 
- hàm mật độ cho biến ngẫu nhiên liên tục 
- mật độ xác suất tại điểm x 
----
3.CDF (cumulative distribution) - hàm phân phối tích luỹ 
- xác suất để biến ngẫu nhiên nhận giá trị <= x 
- áp dụng cho cả biến rời rạc và liên tục
