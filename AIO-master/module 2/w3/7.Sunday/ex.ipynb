- có sự khác nhau giữa công thức xác suất dầy đủ và công thức xác suất hậu nghiệm 
- công thức xs đầy đủ : P(X|Y) = P(X ⋂ Y) / P(Y) 
- công thức xs hậu nghiệm : P(X) ∝ P(Y).P(X|Y).P(Xi|Y) 

- xác suất tiên nghiệm (prior) : là xs xảy ra trước 
- xác suất điều kiện

- sử dụng công thức hàm mật độ xác suất chuẩn gaussian PDF khi dữ liệu đầu vào là liên tục 
- giá trị liên tục là giá trị có thể nhận vô số giá trị khác trong 1 khoảng , thường là các đại lượng đo lường . Trong dữ liệu , nếu mọi thuộc tính là số thực và có thể chia nhỏ vô hạn --> liên tục 

- giá trị rời rạc là giá trị chỉ nhận một số hữu hạn cụ thể , VD : gia đình có 1 2 3 thành viên , không thể 2.5 thành viên 

- nếu là giá trị rời rạc thì sử dụng naive bayes với xs đếm 
- nếu là giá trị liện tục thì có thể sử dụng gaussian naive với hàm mật độ 

### MAP (Maximum A Posteriori)
- MAP (Maximum A Posteriori - Tối đa hóa xác suất hậu nghiệm) là một phương pháp ước lượng tham số trong thống kê Bayesian, kết hợp cả thông tin từ dữ liệu quan sát được (thông qua hàm hợp lý - likelihood) và thông tin tiên nghiệm (prior) về tham số cần ước lượng.

```python
import numpy as np 
```

```python
def create_data_train():
    data = [
        ['Sunny', 'Hot', 'High', 'Weak', 'no'],
        ['Sunny', 'Hot', 'High', 'Strong', 'no'],
        ['Overcast', 'Hot', 'High', 'Weak', 'yes'],
        ['Rain', 'Mild', 'High', 'Weak', 'yes'],
        ['Rain', 'Cool', 'Normal', 'Weak', 'yes'],
        ['Rain', 'Cool', 'Normal', 'Strong', 'no'],
        ['Overcast', 'Cool', 'Normal', 'Strong', 'yes'],
        ['Overcast', 'Mild', 'High', 'Weak', 'no'],
        ['Sunny', 'Cool', 'Normal', 'Weak', 'yes'],
        ['Rain', 'Mild', 'Normal', 'Weak', 'yes']
    ]
    return np.array(data)
```

```python
train_data = create_data_train()
```

```python
def compute_prior_probability(train_data) : 
    y_unique = ['no' , 'yes']    
    prior_probabilty = np.zeros(len(y_unique))
    for i in range(len(y_unique)):
        prior_probabilty[i] = len(np.where(train_data[: , 4] == y_unique[i])[0])/len(train_data)
    return prior_probabilty
```

```python
prior_probabilty = compute_prior_probability(train_data)
prior_probabilty
```
