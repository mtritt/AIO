# rules of probability 

## conditional probability 
- P(A|B) = P(A⋂B)/P(B) 

## multiplication rule 
- P(AB) = P(A).P(B|A) 
- phát biểu : xác suất của B khi A đã xảy ra rồi = xác suất của A và B xảy ra 
- independent : P(AB) = P(A).P(B) 

## bayes's theorem 
- total probability theorem : P(H) = P(H1) + P(H2) + P(H3)
- ==> P(A1H) + P(A2H) + P(A3H) 
- ==> P(A1).P(H1|A1) + P(A2).P(H2|A2) + P(A3).P(H3|A3)
