## What is @app.get('/) ?

- hàm decorator : Decorator trong Python là một công cụ mạnh mẽ dùng để mở rộng, sửa đổi hoặc bao bọc hành vi của các hàm (function) hoặc lớp (class) hiện có mà không làm thay đổi mã nguồn gốc của chúng. Nó hoạt động như một "wrapper", cho phép thêm chức năng (như ghi nhật ký, đo thời gian, kiểm tra quyền) trước hoặc sau khi hàm chính được thực thi. 

### tại sao lại cần decorator? 

- Đăng ký đường dẫn (Routing): Báo cho FastAPI biết "Khi người dùng truy cập vào địa chỉ /predict, hãy chạy hàm này".

- Tái sử dụng code: Thay vì viết đi viết lại code kiểm tra đăng nhập, ghi log (nhật ký) ở mọi hàm, bạn chỉ cần tạo một decorator và "dán" lên trên các hàm đó.

- Tách biệt logic: Giúp hàm chính chỉ tập trung vào việc xử lý dữ liệu (như chạy model AI), còn các việc phụ (như xác thực quyền, định dạng kết quả) để decorator lo.

- khi gõ lệnh uvicorn , FastAPI sẽ quét toàn bộ file code và nếu user sử dụng yêu cầu get/delete/update/.. thì hàm đó sẽ được tìm và thực thi 

- thêm đuôi "/docs" ngay đằng sau để có thể xem và thử swagger UI 

- path operations : are the core of a FastAPI application . They define how clients interact with your API by sepecifying : 
    - the URL path : where the resouce is located 
    - the HTPP method : the action to be performed (get , post , put , delete)
    - the function : the code that will be excuted to handle the request 

- how to send data to the API endpoints : 
    - query parameter 
    - path parameter : nếu có phần path {} thì là path parameter còn không thì là query parameter 
    - request body 

## What is FastAPI 

- là một framwork hiện đại để xây dựng API với python dựa trên các chuẩn python type hints 

- cách sử dụng cơ bản : 
    - khởi tạo : app = FastAPI() 
    - định nghĩa endpoint : sử dụng các decorator như @app.get("/") hoặc post 
    - thực thi : trả về dữ liệu dạng JSON 

## Static file & mounting 

- Static Files là các file tĩnh như ảnh, CSS, JavaScript. Mounting là việc thêm một ứng dụng độc lập vào một đường dẫn cụ thể.

- Giúp API có thể phục vụ cả các file giao diện hoặc dữ liệu cố định mà không cần viết hàm xử lý riêng cho từng file.

- Sử dụng app.mount("/static", StaticFiles(directory="static"), name="static") để ánh xạ thư mục static trong máy tính lên đường dẫn /static của API.
