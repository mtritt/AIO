# Vectorized Linear Regression 

## univerate dataset 

- bộ dữ liệu đơn biến (univariate dataset) là bộ dữ liệu có 1 đầu vào và 1 đầu ra 

## multivariate dataset 

- bộ dữ liệu đa biến (multivariate dataset) là bộ dữ liệu có nhiều hơn một biến số lượng (numeric variable). 

- trong đại số tuyến tính , 1 giá trị được gọi là scalar , nhiều giá trị được gọi là vector , nhiều vector được gọi là matrix 

- dot product (tích vô hướng) : tổng các tích giữa các phần tử tương ứng
