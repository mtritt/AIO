# Basic Linear Regression 

## Simple linear regression 

- tìm một đường thẳng y = ax + b hay yi = wxi + b sao cho độ chính xác cao nhất cho các điểm dữ liệu 
- hay tìm cặp số w và b sao cho đường thẳng gần các điểm dữ liệu nhất 

- error = (y^ - yi)^2 
- loss = tổng(error)
- tìm w và b sao cho loss nhỏ nhất
