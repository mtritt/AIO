# Suport Vector Machine 

- Regression : A task involing predicting a continuous value 
- Classification : A task involing predicting a category


- Suport Vector Machine : A supervised-learning ML algorithm that works by identifying the best hyperplane that differentiates between classes in a feature space

- SVM idea : find the line that best separates the data into classes while maximizing the distances between nearest data points 

- Margin : the distance between the hyperplane and the nearest data points

- Hard Margin SVM : find the line that best separates the data into classes while maximizing the distances between nearest data points

- Hard Margin SVM problem : to avoid , we should use soft margin SVM or allow misclassifications

- Support Vector : draw two lines parallel to the hyperplane and pass through the data points that are closest to the hyperplane

- Decision boundary - Hyperplane : the line that best separates the data into classes
