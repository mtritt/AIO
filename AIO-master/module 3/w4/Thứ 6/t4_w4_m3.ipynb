### AdaBoost : Forest of Stump 

#### Motivation 
- Bạn bắt đầu với một cuốn sách giáo khoa cơ bản (mô hình 1). Sau khi làm bài kiểm tra thử, bạn thấy mình yếu phần Hình học không gian. Ở lần ôn tiếp theo (mô hình 2), bạn dành phần lớn thời gian cho Hình học không gian, gần như bỏ qua các phần khác. Làm bài kiểm tra thử lần 2, bạn thấy mình giỏi Hình học lên nhưng lại quên mất phần Đại số. Lần ôn thứ 3 (mô hình 3) bạn lại tập trung vào Đại số. Cuối cùng, khi đi thi thật, bạn tổng hợp kiến thức từ tất cả các lần ôn (kết hợp có trọng số các mô hình), nhưng bạn nhận thấy rằng chất lượng của mỗi lần ôn là khác nhau nên bạn ưu tiên những lần ôn giúp bạn làm bài tốt hơn.

- Nguyên lý cốt lõi: Điều chỉnh trọng số (weights) của dữ liệu.


- Trong AdaBoost, "Amount of Say" (thường được ký hiệu là αₜ - alpha) chính là trọng số hay tầm quan trọng của một mô hình cơ sở (weak learner) thứ t trong quyết định cuối cùng.
- αₜ = (1/2) * ln((1 - εₜ) / εₜ)

trong đó 
- αₜ: "Amount of say" của mô hình hₜ.
- εₜ: Tổng trọng số (weighted error) của các mẫu bị phân loại sai bởi hₜ. Giá trị này nằm trong khoảng (0, 1).

Vậy 
- Vậy (1 - εₜ) / εₜ là tỷ số giữa độ chính xác và độ sai số.
- Nếu mô hình rất tốt (εₜ rất nhỏ, gần 0): Tỷ số này là một số rất lớn (ví dụ: 0.99/0.01 = 99).
- Nếu mô hình rất tệ (εₜ rất lớn, gần 1): Tỷ số này là một số rất nhỏ (ví dụ: 0.01/0.99 ≈ 0.01).
- Nếu mô hình chỉ như đoán mò (εₜ = 0.5): Tỷ số này bằng 1 (0.5/0.5 = 1).

Tại sao lại cần logarit ?
- Tạo thang đo đối xứng: Logarit giúp biến một tỷ lệ dương (luôn >0) thành một thang đo có thể âm hoặc dương, đối xứng qua điểm εₜ = 0.5.

- Khi εₜ = 0.5 (đoán mò), αₜ = (1/2) * ln(1) = 0. Mô hình vô dụng, không có tiếng nói. Rất logic.

- Khi εₜ < 0.5 (mô hình tốt), (1 - εₜ)/εₜ > 1, ln(...) > 0, do đó αₜ > 0. Mô hình tốt có tiếng nói tích cực.

- Khi εₜ > 0.5 (mô hình tồi tệ, sai nhiều hơn đúng), (1 - εₜ)/εₜ < 1, ln(...) < 0, do đó αₜ < 0. Mô hình tồi có tiếng nói tiêu cực.

- Điều này có ý nghĩa gì? Một mô hình có αₜ âm vẫn có ích! Trong công thức tổng hợp cuối cùng H(x) = sign( Σ αₜ * hₜ(x) ), một αₜ âm có nghĩa là chúng ta sẽ lấy kết quả ngược lại của mô hình tồi đó (-hₜ(x)), và nó lại trở thành một dự đoán tốt. Đây là một điểm rất thông minh của AdaBoost.

Công thức cập nhật trọng số cho mẫu dữ liệu : wᵢ⁽ᵗ⁺¹⁾ = wᵢ⁽ᵗ⁾ * exp(-αₜ * yᵢ * hₜ(xᵢ))

công thức chuẩn hoá : w_norm = w_new / sigma(w_new)

việc tạo ra dataset mới phụ thuộc vào range và sample của new dataset sẽ nằm trong khoảng và ngẫu nhiên 1 số nào đó trong khoảng 

Công thức cuối cùng : 
trong đó : 
- T(m)(x) : weak leaner thứ m 
- a(m) : trọng số 
- I(T(m)(x) == k) : bằng 1 nếu mô hình dự đoán mẫu x thuộc lớp k và ngược lại 
- argmax() : chọn lớp k có tổng trọng số lớn nhất 

- trong các công thức đều nhắc đến Gm(xi) chính là giá trị dự đoán -1 hoặc 1 

#### Cách hoạt động chi tiết:
- Khởi tạo: Gán cho mỗi điểm dữ liệu trong tập huấn luyện một trọng số (weight) bằng nhau (ví dụ: 1/n).

Vòng lặp (cho mô hình thứ m):
- a. Huấn luyện một mô hình cơ sở yếu (thường là một cây quyết định chỉ có 1 nút - decision stump, tức là chỉ hỏi một câu hỏi "Có/Không") trên tập dữ liệu hiện tại, có tính đến trọng số của các điểm dữ liệu.

- b. Đánh giá mô hình: Tính toán lỗi của mô hình này. Mô hình càng sai ở những điểm có trọng số cao thì lỗi càng lớn.

- c. Tính "tiếng nói" cho mô hình: Dựa vào lỗi vừa tính, ta gán cho mô hình này một hệ số alpha. alpha càng lớn khi mô hình càng tốt (lỗi nhỏ). alpha chính là "trọng số" của mô hình trong quá trình bỏ phiếu cuối cùng.

- d. Cập nhật trọng số dữ liệu:
- Giảm trọng số của những điểm mà mô hình dự đoán đúng.
- Tăng trọng số của những điểm mà mô hình dự đoán sai.

- Mục đích là để mô hình tiếp theo sẽ phải "chú ý" nhiều hơn đến những điểm khó (điểm đang bị sai).

- Kết hợp: Dự đoán cuối cùng là tổng có trọng số dựa trên các alpha của tất cả các mô hình. Mô hình nào có alpha cao (tốt) thì sẽ có ảnh hưởng lớn hơn đến kết quả chung.

- Weak Leaners is a Stump 
- AdaBoost combines a lot of Stump 

- sau khi xác định được các sample dự đoán sai của stump thì ta tạo ra 1 datsets mới xuất hiện nhiều các sample dự đoán sai đó để xây dựng các stump sao cho các predict sai đó được fix 

## Gradient Boost clearly explain 

#### Motivation
- Cũng là kỳ thi khó đó. Thay vì tập trung vào các phần cụ thể như AdaBoost, bạn có một cách tiếp cận khác: Lần ôn thứ nhất, bạn học cuốn sách giáo khoa (mô hình 1). Bạn làm bài kiểm tra thử và thấy mình còn sai một số câu. Thay vì tập trung vào một chủ đề, bạn tính xem mình cần cải thiện bao nhiêu điểm ở mỗi câu hỏi để đạt điểm tuyệt đối. Ở lần ôn thứ hai (mô hình 2), bạn không học lại cuốn sách mà học một cuốn sách bài tập nâng cao, chuyên giải quyết chính xác những "lỗ hổng điểm số" đó. Sau lần 2, điểm của bạn đã khá hơn, nhưng vẫn còn một vài lỗ hổng nhỏ. Lần ôn thứ 3 (mô hình 3) bạn học một cuốn sách mỏng để giải quyết nốt những lỗ hổng cuối cùng. Kết quả cuối cùng là tổng điểm của cả 3 lần ôn.

- Nguyên lý cốt lõi: Tối ưu hóa hàm số bằng phương pháp gradient descent. Nó không tập trung vào trọng số dữ liệu, mà tập trung vào "phần dư" (residuals) - tức là sai số của mô hình hiện tại.



- Cách hoạt động chi tiết:

- Khởi tạo: Bắt đầu với một dự đoán rất đơn giản (ví dụ: dự đoán tất cả đều là giá trị trung bình của biến mục tiêu).

Vòng lặp (cho mô hình thứ m):
- a. Tính toán phần dư (residuals). Phần dư chính là đạo hàm của hàm lỗi so với dự đoán hiện tại. Nói đơn giản, nó cho ta biết mô hình hiện tại đang sai lệch bao nhiêu so với giá trị thực tế cho mỗi điểm dữ liệu (residual = giá trị thực - giá trị dự đoán hiện tại).

- b. Huấn luyện một mô hình cơ sở mới (thường là cây quyết định có độ sâu giới hạn) để dự đoán chính xác phần dư này. Tức là mô hình mới sẽ học để dự đoán "cái sai" của mô hình cũ.

- c. Cập nhật mô hình tổng thể: Cộng thêm dự đoán của mô hình mới (thường được nhân với một tỷ lệ học - learning rate nhỏ) vào mô hình tổng thể hiện tại. Mô hình mới = Mô hình cũ + (learning_rate * dự đoán_của_mô_hình_học_phần_dư). Learning rate giúp quá trình học diễn ra chậm và chắc chắn, tránh nhảy quá nhanh đến một kết quả chưa tối ưu.

- Kết hợp: Dự đoán cuối cùng là tổng của dự đoán ban đầu và tất cả các dự đoán phần dư từ các mô hình sau.

- Step 1 : Build 1st tree 
    - chọn giá trị trung bình
- Step 2 : Build 2nd tree 
    - tính phần dư so với kết quả trung bình ban đầu với kết quả thực tế Residual error = y - y^ 
    - để tránh việc bị overfitting thì ta sẽ thêm 1 hệ số alpha là learning rate 
- Step 3 : Build 3rd tree 
    - tính được predicted weight = giá trị trung bình + learning rate * giá trị dự đoán ở 2nd tree 
    - residuals error sẽ giảm dần qua các bước xây dựng 
- lặp lại các bước cho đến khi resuduals error giảm dần về 0 
- ta thường lấy giá trị trung bình là bởi vì tại đó sẽ làm cho hàm đạt giá trị cực tiểu 



Gradient Descent trong không gian hàm :
