## Boosting Method 
- AdaBoost 
- Gradient Boosting 

### AdaBoost (Adaptive Boosting)


- các bước thực hiện thuật toán : 
    - get subset training data 
    - initial sample weight = 1/N 
    - fitting model (tính total error)
    - weight of weak learner 
    - correct predictions : w_new = w * e^-alpha
    - incorrect predictions : w_new = w * e^alpha 
    - normalization
    - sampling probability 

### Gradient Boosting 

- các bước thực hiện thuật toán 
    - initial model predicting 
    - calculating residuals : r1 = y - y^ , r2 = y - y_new 
    - build a decision tree 
    - compute residuals output 
    - update predictions : y_new = y^ + lr(learning_rate) * r^
