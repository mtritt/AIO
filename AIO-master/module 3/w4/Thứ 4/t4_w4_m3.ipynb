# Random forest 

- dummy variables (biến giả) : là các biến chỉ nhận giá trị 0 hoặc 1 được tạo ra để mã hoá dữ liệu định tính (giới tính , vùng miền , màu sắc) thành dạng số để có thể đưa vào mô hình hồi quy hoặc định tính  
- bias (độ chệch) : là sai số đến từ việc học mô hình quá đơn giản , không nắm bắt được quy luật phức tạp của dữ liệu 
- variance (phương sai) : là sai số đến từ việc mô hình quá nhạy cảm với những biến động nhỏ trong dữ liệu , nó học cả những nhiễu loạn 

- high bias : mô hình quá đơn giản thường gây ra UNDERFITTING (mô hình học chưa đủ)
- low bias vs. high variance thường gây ra OVERFITTING (mô hình học vẹt)

----

- ensemble learning : tổng hợp các weak classifier và đưa ra quyết dịnh cuối cùng 
- homogenous approach : hướng tiếp cận đồng nhất tức các cây dùng chung 1 giải thuật rồi tạo ra ensemble leanring đưa ra quyết định cuối cùng 
- heterogeneous approach : tương tự như homogenous approach nhưng mỗi cây lại sử dụng 1 giải thuật khác nhau 

- ensemble được chia thành 3 loại : bagging , boosting , stacking 

### bagging - based method 

- xây dựng N cây rồi từ đó đưa ra quyết định 

### boosting - based method 

- xây dựng cây đầu tiên và các cây tiếp theo khắc phục lỗi của các cây đằng trước 

### stacking - based method 

- xây dựng mỗi cây dựa trên các thuật toán khác nhau

## K - fold cross validation 

- with time series , use : walk forward cross validation 

- K-fold cross validation là cách đánh giá mô hình ML một cách chắc chắn và khách quan hơn so với cách chia dữ liệu thông thường 
- thay vì chia dữ liệu một lần (80 - 20) thì K-fold sẽ chia dữ liệu thành nhiều phần và thực hiện quá trình huấn luyện , kiểm tra nhiều lần để đảm bảo kết quả đánh giá được đảm bảo
