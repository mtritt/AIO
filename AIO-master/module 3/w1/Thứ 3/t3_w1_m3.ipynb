```python
DATASET_PATH = '/Users/macos/AIO/module 3 /w1/Thứ 3/timeseries_daily-minimum-temperatures.csv'
ts_df = pd.read_csv(DATASET_PATH)
ts_df
```

```python
ts_df.shape
```

```python
# ép kiểu thành kiểu date 
ts_df['Date'] = pd.to_datetime(ts_df['Date'])
# cần chuyển qua kiểu dl date vì máy tính đọc là kiểu string 
ts_df = ts_df.set_index('Date')
# index thành Date 
ts_df.head(3)
```

```python
# date format : yy/mm/dd 
ts_df.dtypes #Daily minimum temperatures    object
ts_df['Daily minimum temperatures'] = pd.to_numeric(ts_df['Daily minimum temperatures'] , errors = 'coerce')
ts_df.dtypes
```

```python
# times based indexing 
ts_df.loc['1990-01']
```

```python
# visualizee time-series data 
import matplotlib.pyplot as plt
import seaborn as sns 

sns.set(rc={'figure.figsize':(9, 4)})
col_to_plot = 'Daily minimum temperatures'
ts_df[col_to_plot].plot(linewidth=0.8)
plt.ylabel(col_to_plot)
plt.show()
```

```python
cols_plot = ['Daily minimum temperatures']
ts_df[cols_plot].plot(
    marker='.',
    alpha=0.5,
    linestyle='None',
    figsize=(10, 4),
    subplots=True
)
plt.ylabel('Daily minimum temperatures')
plt.show()
```

- seasonality visualization : Seasonality is a characteristic of a time series in which the data experiences regular and predictable changes that recur every calendar year.

- Tính thời vụ là một đặc trưng của chuỗi thời gian, trong đó dữ liệu trải qua những thay đổi có tính đều đặn và dự đoán được, lặp lại mỗi năm.
