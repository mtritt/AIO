## 1D interpolation

- Phép nội suy (interpolation) trong xử lý dữ liệu là một phương pháp ước tính giá trị của một hàm tại các điểm nằm giữa các điểm dữ liệu đã biết. Nó được sử dụng để "lấp đầy" các khoảng trống trong dữ liệu hoặc tạo ra một hàm liên tục từ dữ liệu rời rạc.



```python
import pandas as pd
import numpy as np 

data = {
    'A': [12.0, 4.0, 5.0, np.nan, 1.0],
    'B': [np.nan, 2.0, 54.0, 3.0, np.nan],
    'C': [20.0, 16.0, np.nan, 3.0, 8.0],
    'D': [14.0, 3.0, np.nan, np.nan, 6.0]
}

df = pd.DataFrame(data)
df
```

```python
df_interpolated = df.interpolate(method='linear', axis=0)
print("\n1. Linear interpolation (theo cột):")
print(df_interpolated)
```

ở đây ta vẫn còn 1 giá trị NaN là do khi thực hiện interpolation linear theo axis = 0 (tức từ trên xuống dưới) , bản chất là nó chỉ nội suy giữa 2 giá trị đã biết , nên giá trị đó không được fill 

công thức linear : y = y1 + (x-x1)*(y2-y1)/(x2-x1)

```python
df_interpolated = df.interpolate(method='linear', axis=0 , limit_direction = 'both')
print("\n2. Linear interpolation (theo cả 2):")
print(df_interpolated)
```

thực hiện việc interpolation theo cả 2 hàng ta được bảng giá trị đã fill full values

### percentile and boxplot 

#### percentile 

- Percentile là giá trị mà tại đó một tỷ lệ phần trăm nhất định của dữ liệu nằm bên dưới giá trị đó.
- ví dụ : Percentile thứ 50 (P50) là giá trị mà 50% dữ liệu nằm dưới nó → chính là median (trung vị).
- công thức : vị trí = P/100 * (n + 1)
- trong đó : P là percentile cần tìm , n là số lượng quan sát 
- P25(Q1) , P50(Q2) , P75(Q3)

#### boxplot

- Boxplot (hay box-and-whisker plot) là biểu đồ trực quan hóa phân phối của dữ liệu dựa trên 5 số tổng kết (five-number summary), sử dụng các percentile để xác định.
- minimum (giá trị nhỏ nhất) : Q1 - 1.5 * IOR 
- maximum (giá trị lớn nhất) : Q3 + 1.5 * IQR 
- IQR (interquartile range) : IQR = Q3 - Q1 
- ouliers (điểm ngoại lai) : [Q1 - 1.5IQR , Q3 + 1.5IQR]

- ascending order 
- find median 
- find Q1 and Q3 
- interquatile IOR = Q3 - Q1

- ouliers : các điểm < Q1 - 1.5*IQR và các điểm > Q3 + 1.5*IQR 
- là các điểm nhiễu
