- ta có thể sử dụng kỹ thuật foward fill , kỹ thuật này ghi nhận được tại thời điểm trước đó làm giá trị thay thế cho giá trị thiếu 

```python
times_sample = pd.date_range('1998-03-10' , '1998-03-14' , freq = 'D')
consum_sample = opsd_daily.loc[times_sample, ['Consumption']].copy()
consum_sample
```

```python
# Convert the data to daily frequency, without filling any missings
consum_freq = consum_sample.asfreq('D')
# tạo ra time series mới với index là mọi ngày trong thời gian 

# Create a column with missings forward filled
consum_freq['Consumption - Forward Fill'] = consum_sample.asfreq('D', method='ffill')

consum_freq
```

## resampling 

- là kỹ thuật dùng để thay đổi tần số biểu diễn của bộ dữ liệu time series , có thể tăng hoặc giảm đi tần số lấy mẫu , ví dụ ta có thể giảm tần số của bộ dữ liệu từ ngày sang tháng 

```python
data_columns = [
    'Consumption',
    'Solar',
    'Wind',
    'Wind+Solar'
]

opsd_daily[data_columns].resample('M').mean() # resample theo tháng 
opsd_daily.head(3)
```

## Rolling windows 

- là việc chuyển thông tin dữ liệu quan trọng time series. Giống downsampling , rolling windows chia dữ liệu thành các time windows (khoảng thời gian tuần , tháng , .. trượt trên các mẫu dữ liệu theo ngày) và dữ liệu đó được tổng hợp với các hàm mean() , median() , sum() , ... 
- tuy nhiên , không giống với downsampling khi mà dữ liệu không overlap(chồng chéo) và output luôn có tần số thấp hơn input , rolling windows overlap và gom thành những dữ liệu có cùng tần số , vì thees time series được chuyển cùng tần số với time series gốc 

```python
opsd_7d = opsd_daily[data_columns].rolling(7 , center = True).mean()
opsd_7d.head(10)
```

## trends 

```python
import matplotlib.dates as mdates

# The min_periods=360 argument accounts for a few isolated missing days in the
# wind and solar production time series
opsd_365d = opsd_daily[data_columns].rolling(window=365, center=True, min_periods=360).mean()

# Plot daily, 7-day rolling mean, and 365-day rolling mean time series
fig, ax = plt.subplots()
ax.plot(opsd_daily['Consumption'], marker='.', markersize=2, color='0.6',
        linestyle='None', label='Daily')
ax.plot(opsd_7d['Consumption'], linewidth=2, label='7-d Rolling Mean')
ax.plot(opsd_365d['Consumption'], color='0.2', linewidth=3,
        label='Trend (365-d Rolling Mean)')
# Set x-ticks to yearly interval and add legend and labels
ax.xaxis.set_major_locator(mdates.YearLocator())
ax.legend()
ax.set_xlabel('Year')
ax.set_ylabel('Consumption (GWh)')
ax.set_title('Trends in Electricity Consumption')
plt.show()
```
