## Regularization

- Regularization (Chính quy hoá) là một trong những kĩ thuật quan trọng nhất trong Machine learning và Deeplearning nhằm giải quyết vấn đề Overfitting 

1. cơ chế hoạt động : 
- mục tiêu là tìm ra bộ trọng số (weights) sao cho hàm mất mát (loss function) đạt giá trị nhỏ nhất 
    - $$J(w) = \text{Loss}(w) + \lambda \cdot R(w)$$
        - $J(w)$: Hàm mất mát mới (đã có Regularization).
        - $\text{Loss}(w)$: Hàm mất mát gốc (ví dụ: Mean Squared Error trong Hồi quy).
        - $R(w)$: Thành phần phạt, phụ thuộc vào độ lớn của các trọng số $w$.
        - $\lambda$ (Lambda): Siêu tham số do bạn tự chọn, quyết định "mức độ phạt". Nếu $\lambda = 0$, mô hình quay về trạng thái không có Regularization. Nếu $\lambda$ quá lớn, mô hình sẽ bị Underfitting (Chưa khớp).
        

2. CÁCH XGBOOST HOẠT ĐỘNG CHI TIẾT 

    - đối với bài toán Regression (Hồi quy)

        1. dự đoán ban đầu : thuật toán bắt đầu với một dự đoán mặc định là 0.5 
        2. tính toán phần dư : Residual = output - predict 
        3. tính điểm tương đồng (Similarity Score - SC) : 
            - $$SC = \frac{(\sum \text{Residual})^2}{m + \lambda}$$
            - Trong đó $m$ là số lượng mẫu (phần dư) trong lá, và $\lambda$ là tham số Regularization.
        4. tách nhánh : thuật toán thử các điểm tách khác nhau , tính SC cho lá node trái và node phải , sau đó đánh giá độ hiệu quả bằng Gain 
            - $$Gain = \text{Left SC} + \text{Right SC} - \text{Root SC}$$
            - có thể prunning bằng cách : chọn trọng số gamma , tính diff = gain - gamma 
                - nếu diff > 0 , không cắt nhánh 
                - nếu diff < 0 , cắt nhánh 
                ==> cũng có thể hiểu là tại vị trí gain càng lớn thì lượng thông tin càng nhiều nên không thể cắt 
        5. tính giá tri đầu ra của lá 
            - $$\text{Output Value} = \frac{\sum \text{Residual}}{\text{Number of Residuals} + \lambda}$$
        6. cập nhật dự đoán mới 
            - Dự đoán mới bằng dự đoán cũ cộng với Output Value nhân với Tốc độ học 
            - ($\alpha$): $\text{Prediction} + \alpha \times \text{Output}$.+2
        

3. CÁCH XGBOOST HOẠT ĐỘNG CHI TIẾT (đối với bài toán phân loại)

    - đối với bài toán Classification (Phân loại)
        1. Dự đoán ban đầu: Bắt đầu bằng xác suất mặc định (ví dụ: xác suất 0.5)
        2. Tính phần dư: Hiệu số giữa nhãn thực tế (1 hoặc 0) và xác suất dự đoán.
        3. Công thức SC và Output Value thay đổi: 
            - $$SC = \frac{(\sum \text{Residual})^2}{\sum [\text{previous probability} \times (1 - \text{previous probability})] + \lambda}$$
            - $$\text{Output Value} = \frac{\sum \text{Residual}}{\sum [\text{previous probability} \times (1 - \text{previous probability})] + \lambda}$$
        4. Chuyển đổi qua lại bằng Log(odds): XGBoost sử dụng Log(odds) để cộng dồn các giá trị dự đoán từ cây. 
            - $\log(\text{odds}) = \log(\frac{p}{1-p})$.
        5. Cập nhật xác suất: Sau khi cộng giá trị $\text{Output}$ (đã nhân $\alpha$) vào $\log(\text{odds})$ cũ, thuật toán biến đổi ngược lại thành xác suất bằng hàm Sigmoid:$$Probability = \frac{e^{\log(\text{odds})}}{1 + e^{\log(\text{odds})}}$$
            

- vai trò của các tham số : 

    - $\lambda$ (Lambda - Regularization): Tham số điều chuẩn. Khi $\lambda > 0$, giá trị Similarity Score và Output Value sẽ nhỏ đi. Việc tăng $\lambda$ ép giá trị dự đoán tiến dần về 0, giúp dễ dàng cắt tỉa cây hơn và ngăn chặn mô hình học quá mức (Overfitting).

    - $\gamma$ (Gamma - Pruning): Tham số cắt tỉa cây. Sau khi xây dựng xong toàn bộ cây , thuật toán xét mức độ hiệu quả: Difference = Gain - Gamma. Nếu Difference < 0, nhánh đó sẽ bị cắt bỏ. XGBoost vẫn có thể tự cắt tỉa ngay cả khi $\gamma = 0$ nếu Gain ra âm.

    - $\alpha$ (Learning Rate): Được ký hiệu là tỷ lệ học, dùng để nhân với Output Value của cây mới trước khi cộng vào dự đoán tổng. Điều này làm giảm mức độ ảnh hưởng của một quan sát cá thể lên kết quả dự đoán chung.+2

    - Cover (Min Child Weight): Là ngưỡng tối thiểu để tồn tại một lá.Hồi quy: 
        - Cover bằng số lượng Residuals trong lá.Phân loại: 
        - $\text{Cover} = \sum [\text{previous probability} \times (1 - \text{previous probability})]$.
        - Giá trị Cover tối thiểu mặc định thường là 1.
