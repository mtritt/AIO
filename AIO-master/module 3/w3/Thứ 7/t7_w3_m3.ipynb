## Apache hadoop 

- hadoop is an open source platform used to store and process large datasets on clusters of computers 
- Components : Hadoop Distributed File System (HDFS) and MapReduce 

## Apache Spark 

- Spark is a data processing tool that is faster and more flexible than Haddop 
- It uses in-memory data processing , which significantly speeds up computations 

- Spark lưu trữ ở Ram nên tốc độ sẽ nhanh hơn Hadoop 
- Spark kết hợp được với nhiều database khác
