# Exercise : Decision Tree 

- Decision Tree (Cây quyết định) : là phương pháp học có giám sát không tham số dựa vào ý tưởng xây dựng mô hình dạng cây để xấp xỉ dữ liệu huấn luyện . Cấu trúc cây quyết định được mô tả :
- cây có các node là feature của dữ liệu 
- các đường nối là giá trị của thuộc tính 
- các node lá đại diện cho các kết quả có thể của mô hình 

- để xây dựng cây quyết định , ta sử dụng thuật toán Iterative Dichotomiser (ID3) xây dựng từ trên xuống dựa theo một số độ đo xác định thuộc tính nào sẽ là node gốc và các thuộc tính node con sẽ là gì 
- Một số độ đo quan trọng thường được sử dụng để đánh giá là Gini Impurity hoặc Entropy / Information Gain cho bài toán classification và Variance / Sum of Squared Errors cho bài toán regression 


## Decision Tree for Classification 

- độ đo gini có thể xác định bằng công thức : gini(D) = n1/n * gini(D1) + n2/n * gini(D2) 
- trong đó : gini(Di) = 1 - sum(1-j)(pj)^2
- Gini ∈ [0, 0.5] cho bài toán 2 lớp
- Gini ∈ [0, 1-1/k] cho bài toán k lớp (với k ≥ 2)
----
- công thức gini(D) chỉ được áp dụng khi phân tách bộ dữ liệu để so sánh với các thuộc tính khác 

- độ đo entropy tương tự như độ đo gini nhưng khác công thức trong đó 
- entropy = -sum(1-j)(pj * log2(pj))
- Entropy ∈ [0, 1] (khi dùng log cơ số 2)

- Information Gain được xđ bằng công thức : Gain(D) = 1 - Entropy(D) 
- Gain ∈ [0, Entropy_ban_đầu]

- để xác định được thuộc tính quan trọng với cây quyết định chúng ta có thể chọn Entropy hoặc Gini thấp nhất , Gain cao nhất 

- khi sử dụng công thức gini(D) ta sẽ phải tính gini của thuộc tính và so sánh với cả feature quyết định xem là 0 hay 1 

## Decision Tree for Regression 

- với bài toán regression để chọn được thuộc tính phân chia dữ liệu và xây dựng cây , chúng ta sử dụng độ đo SSE . Độ do sum squared error(SSE) 
- SSE(D) = SSE(D1) + SSE(D2)
- SSE(Di) = sum(1-ni)(xj− xi)^2

- trong đó : xi ngang là giá trị trung bình của tập Di
