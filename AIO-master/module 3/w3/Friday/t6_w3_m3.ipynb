# Decision Tree for Regression 

## Gini vs entropy 

- chỉ có 2% kết luận khác nhau giữa gini và entropy 
- entropy chậm hơn so với gini 

- cả 2 thuật toán đều đươc sử dụng để đo độ impurity , giá trị càng cao -> càng nhiều class và độ hỗn loạn càng lớn 

- maximum của entropy dựa vào số lượng class và được quyết định bởi công thức 
- entropy(max) = log2(n)

- maximum của gini cũng tương tự như entropy 
- gini(max) = (n-1)/n

# Regression tree 

## Motivation 

- giá trị bài toán là rời rạc ta có thể sử dụng entropy & gini để tính toán 
- nhưng đối với bài toán là giá trị liên tục ta sẽ phải sử dụng cách khác 

- sử dụng Sum Square Error (SSR) để làm thước đo 

## Overfitting problem 

- underfit khi error cao với cả trainning set và test set 
- overfit khi error cao với test set nhưng thấp với training set 

- high bias thì error cao tức kết quả thực tế khác xa với kết quả nhận được 
- variance : là sự khác biệt khi sử dụng 2 dataset khác nhau 

- thông thường , chúng ta sẽ giới hạn số node (observation) tối đa để thực hiện tách nhánh là 20 để tránh overfitting
