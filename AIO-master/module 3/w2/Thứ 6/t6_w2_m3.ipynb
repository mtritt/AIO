# K - means 

```python
import numpy as np 
```

```python
X = np.array([1.4 , 1.5 , 1.0 , 3.0 , 3.8 , 3.1])
X
```

```python
C = X[:2]
C
```

```python
np.abs(X - 1.4)
```

```python
np.abs(X - 1.0)
```

```python
D = np.abs(X - C)
```

```python
""" vì không thể thực hiện trừ đối với 2 array không cùng shape """
#data = X.reshape(6 , 1)
data = X.reshape(-1 , 1) # tự động điền số chiều vào -1 
data 
```

```python
D = np.abs(data - C) 
D 
```

```python
L = np.argmin(D , axis = 1)
L 
```

```python
X[L == 0].mean()
```

```python
X[L == 1].mean() 
```

```python
[X[L == i].mean() for i in range(2)]
```

```python
k = 2 
# 1.randomly select centroids
print('X\n' , X)
centroids = X[:k]
print('centroids\n' , centroids)

for i in range(10) : 
    # 2.compute distance
    distance = np.abs(X.reshape(-1 ,1) -  C)
    print('distance\n' , distance)

    # 3.compute labels 
    labels = np.argmin(distance , axis = 1)
    print('labels\n' , labels)

    # 4.update centroids
    centroids_new = np.array([X[labels == i].mean() for i in range(k)])
    print(centroids_new)

    # check 
    if np.all(centroids_new == centroids): 
        break

    centroids = centroids_new  

print(centroids)
```
