### what is the computer vision 
- object detection = classification + localization 
- input : an image 
- ouput : a list of predicted bounding boxes in the form of (xmin , ymin , xmax , ymax , confidence score) 
- confidence score : a value represents how 'confident' the model is to assign a class to the corresponding object 

- problem statement : given a helmet safety dataset , train a YOLOv10 model to detect whether people in an image are wearing a helmet or not ? 

## introduce YOLOv10 

### training command and flags meaning
- img : the size of training image 
- batch : number of data samples to read in 1 epoch 
- epochs : number of iteration (number of times model "see" data training)
- data : path to .yaml file 
- weights : path to pretrained model 

### basic training type 
- training : start with randomized weights and train the model on a new stack 
- pre-training : start with a randomized weights and train the model on a very large dataset - wide range of tasks
- fine-tuning : start with pre-trained weights and train the model on a new task
