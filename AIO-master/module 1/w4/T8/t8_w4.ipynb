- AI models that were trained on a very large corpus of text . This made them capable of performing various NLP tasks with high precision 
- LLMs are often pretrained on a vast majority of data and designed to be adapable to a wide variety of tasks (Foundation models)

# RAG (Retrieval-Augmented Generation)
- là kỹ thuật giúp các mô hình ngôn ngữ lớn trả lời các câu hỏi chính xác và đáng tin cậy hơn bằng cách truy xuất thông tin từ một nguồn kiến thức bên ngoài trước khi đưa ra câu trả lời 
- vector database : là database chứa các vector , các vector lại chứa các con số chúng chính là các tài liệu , chuyển text -> vector ,.... 

- retriever pipeline : 
file.pdf -> load file -> text splitter -> vectorization -> vector database 

- langchain : a framework for developing applications powered by LLMs 
- langchain components : documents loaders and utils 

- Chunking is the process of breaking down a large piece of information (like a long document, video, or audio file) into smaller, more manageable, and semantically meaningful segments or "chunks."
- giúp máy tính xử lý thông tin hiệu quả hơn . Các mô hình AI thường có giới hạn về độ dài văn bản đầu vào (context window) . Chunking giúp chia nhỏ kiến thức để phù hợp hơn với giới hạn này 
- chất lượng chunk ảnh hưởng trực tiếp đến hiệu suất 

- embedding : là kỹ thuật biến đổi các đối tượng đầu vào (văn bản , hình ảnh , video ,...) thành một dãy số (vector) trong không gian nhiều chiều 
- mỗi vector này là một "biểu diễn số" của đối tượng gốc . Các đối tượng có ý nghĩa tương tự nhau sẽ nằm gần nhau trong khoảng này 
- giúp máy tính hiểu được ngữ nghĩa của dữ liệu 

- retriever : là thành phần có nhiệm vụ tìm kiếm và truy xuất các thông tin có liên quan nhất từ cơ sở dữ liệu để trả lời 1 query hiệu quả 
- trong RAG , retriever thường tìm trong CSDL các chunk đã được embedding 


--> chunk : chuẩn bị 
--> embedding : số hoá dl 
--> retriever : tìm kiếm và đưa ra kết quả 

- AI học bằng cách "tiêu thụ" một dữ liệu văn bản khổng lồ , với mỗi một từ ngữ được học sẽ có một con số đặc trưng riêng và ngữ cảnh sẽ định nghĩa dần dần theo đó các số có khoảng cách gần nhau thì sẽ có ý nghĩa gần nhau
